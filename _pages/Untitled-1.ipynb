```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
```

```python
dx = .2
x0 = 0
x1 = 2
Nx = int((x1 - x0)/dx)
x_grid = np.linspace(x0, x1, Nx)
y = np.zeros_like(x_grid)

# Initial condition
y[0] = 1

# Forward solution
def forward_euler(x_grid, y, dx):
    '''Forward solver of dy/dx=-x using the Forward Euler method.'''
    for i in range(len(x_grid) - 1):
        y[i+1] = (1-dx)*y[i]
    return y

# Analytical solution
y_a = np.exp(-x_grid)

plt.plot(x_grid, y_a, label='Analytical solution')
plt.plot(x_grid, forward_euler(x_grid, y, 0.2), label='Forward Euler')
plt.title('Forward Euler versus Analytical')
plt.legend()
```

Comparison of Forward Euler solution at $x=2$ for different values of $\Delta x$ 

```python
dxs = np.logspace(start=-1,stop=-4, num=4, base=2)

fig, ax = plt.subplots(nrows=2, figsize=[8,10])

for dx in dxs:
    Nx = int((x1 - x0)/dx)
    x_grid = np.linspace(x0, x1, Nx)
    y = np.zeros_like(x_grid)

    # Initial condition
    y[0] = 1
    y_fe = forward_euler(x_grid, y, dx)
    ax[0].plot(x_grid, y_fe, label=dx);
    ax[1].scatter(dx, np.abs(y_fe[-1] - np.exp(-2)))
# Analytical solution
y_a = np.exp(-x_grid)
ax[0].plot(x_grid, y_a, label='Analytical');
ax[0].legend();
```
